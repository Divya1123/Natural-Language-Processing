```python
corpus = """Hello, welcome to my NLP repository. 
Please install NLTK library to run these codes.
"""
corpus
```

```python
from nltk.tokenize import sent_tokenize 
documents= sent_tokenize(corpus)
print(documents)
```

```python
from nltk.tokenize import word_tokenize
word_tokens = word_tokenize(corpus)
print(word_tokens)
```

```python
for sentences in documents:
    print(word_tokenize(sentences))
```

## Stemming

```python
# PorterStemmer technique
from nltk.stem import PorterStemmer
stemming = PorterStemmer()

for word in word_tokens:
    print(word+"----->"+stemming.stem(word))
```

$ - means end of string in regex<br>
^ - means start of the string

```python
# RegexpStemmer - takes regular expression and remove any prefix or suffix that matches it.
from nltk.stem import RegexpStemmer
reg_stemmer = RegexpStemmer('ing$| s$| able$', min=4)
```

```python
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer = RegexpStemmer('^ing| s$| able$', min=4)
reg_stemmer.stem('ingeating')
```

```python
reg_stemmer = RegexpStemmer('ing| s$| able$', min=4)
reg_stemmer.stem('ingeating')
```

```python
# Snowball Stemmer
from nltk.stem import SnowballStemmer
snow_stemmer = SnowballStemmer('english')

for word in word_tokens:
    print(word+"--->"+snow_stemmer.stem(word))
```

## Stopwords

```python
nltk.download('stopwords')
```

```python
from nltk.corpus import stopwords
stopwords.words('english')
```

```python
paragraph = """Gross Domestic Product (GDP) is the total monetary value of all finished goods and services produced within a country's borders during a specific period, usually a year. 
It serves as the most common measure of an economy's size and health, with growth rates indicating economic progress.
"""
```

```python
sentences = sent_tokenize(paragraph)
sentences
```

```python
stopword = set(stopwords.words('english'))
for i in range(len(sentences)):
    words = word_tokenize(sentences[i])

    words = [stemming.stem(word) for word in words if word not in stopword]
    sentences[i] = ' '.join(words)

sentences
```
